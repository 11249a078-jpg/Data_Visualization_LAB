```r
import matplotlib.pyplot as plt

subjects = ["python", "java", "c++", "ai"]
students = [40, 35, 25, 45]

plt.bar(subjects, students)

plt.title("Bar Chart")
plt.xlabel("Subjects")
plt.ylabel("Number of Students")

plt.show()
```

```r
import matplotlib.pyplot as plt
months=["jan","feb","march","april","may"]
sales=[100,120,150,130,170]
plt.plot(months,sales)
plt.title("line chart")
plt.show()
```

```r
import matplotlib.pyplot as plt

labels = ["cse", "it", "ece", "eee"]
values = [120, 90, 80, 60]

plt.pie(values, labels=labels, autopct="%1.1f%%")



plt.show()
```

```r
import matplotlib.pyplot as plt
x=[10,20,30,40,50]
y=[15,25,35,45,55]
plt.scatter(x,y)
plt.show()
```

```r
import matplotlib.pyplot as plt
marks=[55,60,65,70,75,80,85,90]
plt.hist(marks)
plt.show()
```

```r
import matplotlib.pyplot as plt

months = ["jan", "feb", "mar", "apr", "may"]
sales = [100, 150, 130, 170, 200]

plt.plot(months, sales, marker='o', color='blue')

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

```r
X <- c(1, 2, 3, 4, 5)
Y <- c(10, 15, 12, 18, 20)

plot(X, Y,
     type = "o",
     col = "blue",
     main = "Line Graph",
     xlab = "X values",
     ylab = "Y values")
```

```r
X <- c(1, 2, 3, 4, 5)
Y <- c(10, 15, 12, 18, 20)

barplot(Y,
        names.arg = X,
        col = "skyblue",
        main = "Bar Chart",
        xlab = "X values",
        ylab = "Y values")
```

```r
X <- c(1, 2, 3, 4, 5)
Y <- c(10, 15, 12, 18, 20)

plot(X, Y,
     pch = 19,
     col = "red",
     main = "Scatter Plot",
     xlab = "X values",
     ylab = "Y values")
```

```r
X <- c(1, 2, 3, 4, 5)
Y <- c(10, 15, 12, 18, 20)

pie(Y,
    labels = paste("X", X),
    col = rainbow(length(Y)),
    main = "Pie Chart")

```

```r
y <- c(9, 17, 14, 18, 22, 27)

hist(y,
     col = "lightgreen",
     border = "black",
     main = "Histogram",
     xlab = "X values",
     ylab = "Frequency")
```
